# 03 — Reproduction results

Load a trained checkpoint, run evaluation, and produce the comparison table.

```python
import torch, pandas as pd
from omegaconf import OmegaConf
from glaucoma_rdcnn.models import build_rdcnn
from glaucoma_rdcnn.engine.evaluator import Evaluator
from glaucoma_rdcnn.data import build_dataloader
from glaucoma_rdcnn.utils.checkpoint import load_checkpoint

CFG = OmegaConf.load('../configs/default.yaml')
CFG.data = OmegaConf.load('../configs/data/drishti_gs.yaml')
CFG.model = OmegaConf.load('../configs/model/rdcnn.yaml')
device = 'cuda' if torch.cuda.is_available() else 'cpu'
model = build_rdcnn(CFG.model).to(device)
# load_checkpoint('../outputs/<run>/best.ckpt', model=model, map_location=device)
loader = build_dataloader(CFG.data, 'test')
metrics = Evaluator(CFG).evaluate(model, loader, device=device)
pd.Series(metrics)
```
