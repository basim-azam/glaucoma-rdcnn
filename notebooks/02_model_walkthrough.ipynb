# 02 — Model walkthrough

Build R-DCNN from configs, run a forward pass, inspect intermediate shapes.

```python
import torch
from omegaconf import OmegaConf
from glaucoma_rdcnn.models import build_rdcnn
cfg = OmegaConf.load('../configs/model/rdcnn.yaml')
cfg.backbone.pretrained = False    # avoid downloading in a notebook
model = build_rdcnn(cfg).eval()
x = torch.randn(1, 3, 800, 800)
with torch.no_grad(): out, _ = model(x)
print('OD boxes:', out[0]['od_boxes'].shape)
print('OC boxes:', out[0]['oc_boxes'].shape)
```

```python
# Backbone shapes
feats = model.backbone(x)
print({k: v.shape for k, v in feats.items()})
```
