# 01 — Data exploration

Walk through the preprocessed dataset, plot a few samples with their OD/OC masks overlaid, and check class balance / image statistics.

```python
import json, cv2, numpy as np, matplotlib.pyplot as plt
from pathlib import Path
DATASET = Path('../data/drishti_gs')   # change to ../data/rim_one_v3
splits = json.loads((DATASET / 'splits/official.json').read_text())
print({k: len(v) for k, v in splits.items()})
```

```python
def show_sample(stem):
    img = cv2.cvtColor(cv2.imread(str(DATASET / f'images/{stem}.png')), cv2.COLOR_BGR2RGB)
    od = cv2.imread(str(DATASET / f'masks/od/{stem}.png'), 0)
    oc = cv2.imread(str(DATASET / f'masks/oc/{stem}.png'), 0)
    fig, axes = plt.subplots(1, 3, figsize=(12, 4))
    axes[0].imshow(img); axes[0].set_title(stem)
    axes[1].imshow(od, cmap='gray'); axes[1].set_title('OD mask')
    axes[2].imshow(oc, cmap='gray'); axes[2].set_title('OC mask')
    for a in axes: a.axis('off')
show_sample(splits['train'][0])
```
